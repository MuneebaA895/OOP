**6. Magic Methods / Operator Overloading**

You do not call magic methods directly. Python calls them for you, automatically, in certain situations - like when an object is created, printed, or compared to another object.

```python
# 1. Use __repr__() to describe the object:

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __repr__(self):
    return f"Person('{self.name}', {self.age})"

p1 = Person("Tobias", 25)
print(p1)
```

```python
# 2. Add a __eq__() method that defines that two objects are equal if the values are equal:
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __eq__(self, other):
    return self.name == other.name and self.age == other.age

p1 = Person("Linus", 30)
p2 = Person("Linus", 30)

print(p1 == p2)
```

```python
# 3.  Use __add__() to add the age of two Person objects together:
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __add__(self, other):
    return self.age + other.age

p1 = Person("Emil", 22)
p2 = Person("Tobias", 19)

print(p1 + p2)
```

```python
# 4. The __len__() method controls what the built-in len() function returns for your object.
class Company:
  def __init__(self, employees):
    self.employees = employees

  def __len__(self):
    return len(self.employees)

c1 = Company(["Emil", "Tobias", "Linus"])

print(len(c1))
```

```python
# 5. The __lt__() method ("less than") controls what the < operator does for your own objects
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __lt__(self, other):
    return self.age < other.age

p1 = Person("Emil", 22)
p2 = Person("Tobias", 19)

print(p1 < p2)
```

```python
# 6. The __contains__() method controls what the in operator checks for your own objects.
class Company:
  def __init__(self, employees):
    self.employees = employees

  def __contains__(self, name):
    return name in self.employees

c1 = Company(["Emil", "Tobias", "Linus"])

print("Emil" in c1)
```

```python
# 7. The __call__() method lets an object be called like a function, using object() syntax.
class ClickCounter:
  def __init__(self):
    self.clicks = 0

  def __call__(self):
    self.clicks += 1
    return self.clicks

button_clicks = ClickCounter()

print(button_clicks())
print(button_clicks())
print(button_clicks())
```

**7. Python Inheritance**

```python
class Person:
  def __init__(self, name, age):
    self.name=name
    self.age=age

Ali=Person("Ali",30)
print(f"{Ali.name}")
print(f"{Ali.age}")
```

```python
# Creating a child class
class Student(Person):
  pass

# Creating an object of Student child Class
s1 = Student("Ahmed",15)
print(f"{s1.name}")
print(f"{s1.age}")
```

```python
# When you add the __init__() function, the child class will no longer inherit the parent's __init__() function.
# Creating a child class
class Student(Person):
  def __init__(self, city):
    self.city=city

# Creating an object of Student child Class
s1 = Student("Karachi")

# The child's __init__() function overrides the inheritance of the parent's __init__() function.
# print(f"{s1.name}")  gives error
# print(f"{s1.age}")   gives error
print(f"{s1.city}")
```

Python also has a super() function that will make the child class inherit all the methods and properties from its parent:

```python
class Insect:
  def __init__(self, legs,wings):
    self.legs=legs
    self.wings=wings

class Butterfly(Insect):
  def __init__(self, name, legs, wings):
    super().__init__(legs,wings)
    self.name=name
```

```python
Butter = Butterfly("butterfly","yes","yes")
print(f"The insect is {Butter.name}")
print(f"It has wings? :{Butter.wings}")
print(f"It has legs?: {Butter.legs}")
```

```python
# If you add a method in the child class with the same name as a
# function in the parent class, the inheritance of the parent method will be overridden.
```

**8. Class Polymorphism**

```python
# Polymorphism is often used in Class methods, where we can have multiple classes with the same method name.
class Car: # Class 1
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Drive!")

class Boat: # Class 2
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Sail!")

class Plane: # Class 3
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
  x.move()
```

```python
# What about classes with child classes with the same name? Can we use polymorphism there? YES!
class Vehicle:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Move!")

class Car(Vehicle): # inherits move method from parent class
  pass

class Boat(Vehicle):
  def move(self):  # overrides the parent move method
    print("Sail!")

class Plane(Vehicle):
  def move(self):  # overrides theparent move method
    print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
  print(x.brand)
  print(x.model)
  x.move()
```

**9. Encapsulation**

```python
class BankAccount:
  def __init__(self, accountName):
    self.accountName=accountName
    self.__password= "568YYS"


a1=BankAccount("Aleena")
print(a1.accountName)
# print(a1.__password) will cause an error
```

```python
# Getter and Setter methods to access private data members

class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age # private data member

  # getter method
  def get_age(self):
    return self.__age

  # setter method
  def set_age(self, age):
    if age > 0:
      self.__age = age
    else:
      print("Age must be positive")

p1 = Person("Tobias", 25)
print(p1.get_age())

p1.set_age(26)
print(p1.get_age())
```

```python
# Private methods
class Calculator:
  def __init__(self):
    self.result = 0

  def __validate(self, num):
    if not isinstance(num, (int, float)):
      return False
    return True

  def add(self, num):
    if self.__validate(num):
      self.result += num
    else:
      print("Invalid number")

calc = Calculator()
calc.add(10)
calc.add(5)
print(calc.result)
# calc.__validate(5) # This would cause an error
```

**10. Inner Classes**

```python
class University:
    def __init__(self, name):
        self.name = name

    class Department:
        def __init__(self, dept_name):
            self.dept_name = dept_name


# Create an outer class object
uni = University("Mair University")

# Create an inner class object
dept = University.Department("Computer Science")

print(uni.name)
print(dept.dept_name)
```

```python
class University:
    def __init__(self, name):
        self.name = name

    class Department:
        def __init__(self, university):
            self.university = university

        def display(self):
            print(self.university.name)


uni = University("Mair University")
dept = University.Department(uni)

dept.display()
```

The main difference b/w inheritance and inner classes is that inheritance represents an "is-a" relationship, while an inner class is used to organize closely related classes. Inner class: A "belongs-to" relationship

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model
    self.engine = self.Engine()

  class Engine:
    def __init__(self):
      self.status = "Off"

    def start(self):
      self.status = "Running"
      print("Engine started")

    def stop(self):
      self.status = "Off"
      print("Engine stopped")

  def drive(self):
    if self.engine.status == "Running":
      print(f"Driving the {self.brand} {self.model}")
    else:
      print("Start the engine first!")

car = Car("Toyota", "Corolla")
car.drive()
car.engine.start()
car.drive()
```

```python
class Computer:
  def __init__(self):
    self.cpu = self.CPU()
    self.ram = self.RAM()

  class CPU:
    def process(self):
      print("Processing data...")

  class RAM:
    def store(self):
      print("Storing data...")

computer = Computer()
computer.cpu.process()
computer.ram.store()
```
